this project aims to see the efficacy of linear regtressioin on data following different distributions and on data with different error distributions and max errors and then go on to explore how GLMs improve the predictions.

```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
x1=np.random.normal(0,5,500)
x2=np.random.normal(0,5,500)
x=[x1,x2]
y=-25*x1+3*x2**2 + np.random.normal(0,15,500)
plt.scatter(x2,y)
plt.show()
plt.scatter(x1,y)
plt.show()
```

```python
x=[x1,x2]
linReg=LinearRegression()
linReg.fit(x1,y)
```

```python
# The learned intercept (should be close to 4)
print(f"Learned Intercept: {linReg.intercept_}")

# The learned coefficient for X (should be close to 3)
print(f"Learned Coefficient: {linReg.coef_}")
```
